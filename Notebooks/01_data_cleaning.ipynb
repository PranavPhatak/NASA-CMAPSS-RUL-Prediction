```python
import pandas as pd
import numpy as np
```

```python
columns = [
    "unit_id",
    "cycle",
    "setting_1",
    "setting_2",
    "setting_3",
]

columns += [f"sensor_{i}" for i in range(1, 22)]
```

```python
df = pd.read_csv("../CMAPSSData/Raw/train_FD001.txt", names=columns, sep=r"\s+", header=None)
df.head()
```

```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

```python
round(df.describe(), 2)
```

```python
df.isnull().sum()
```

```python
print("Duplicates Rows:", df.duplicated().sum())
```

```python
sensor_cols = [col for col in df.columns if col.startswith("sensor")]

df[sensor_cols].nunique().sort_values()
```

```python
cycle_check = df.groupby("unit_id")["cycle"].agg(["min", "max", "count"])

cycle_check["expected_count"] = (
    cycle_check["max"] - cycle_check["min"] + 1
)

cycle_check["missing_cycles"] = (
    cycle_check["expected_count"] - cycle_check["count"]
)

cycle_check.head()
```

```python
np.isinf(df.select_dtypes(include=np.number)).sum().sum()
```

```python
df.duplicated(subset=["unit_id", "cycle"]).sum()
```

```python
constant_sensor = ['sensor_1', 'sensor_5', 'sensor_10', 'sensor_16', 'sensor_18', 'sensor_19']

df = df.drop(columns=constant_sensor)
```

```python
output_path = "../CMAPSSData/processed/train_FD001_cleaned(removed_the_sensor).csv"

df.to_csv(output_path, index=False)

print(f"Cleaned data saved to: {output_path}")
print(f"Shape: {df.shape}")
```
