```python
import joblib
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
WITH_CONSTANT_TRAIN = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_training(not_removed_the_sensor).csv"
WITH_CONSTANT_VAL = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_validation(not_removed_the_sensor).csv"

WITH_CONSTANT_TRAIN_SCALED = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_training_scaled(not_removed_the_sensor).csv"
WITH_CONSTANT_VAL_SCALED = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_validation_scaled(not_removed_the_sensor).csv"

WITH_CONSTANT_SCALER = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_scaler(not_removed_the_sensor).pkl"

WITHOUT_CONSTANT_TRAIN = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_training(removed_the_sensor).csv"
WITHOUT_CONSTANT_VAL = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_validation(removed_the_sensor).csv"

WITHOUT_CONSTANT_TRAIN_SCALED = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_training_scaled(removed_the_sensor).csv"
WITHOUT_CONSTANT_VAL_SCALED = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_validation_scaled(removed_the_sensor).csv"

WITHOUT_CONSTANT_SCALER = "../CMAPSSData/Processed/train_FD001_cleaned_added_RUL_scaler(removed_the_sensor).pkl"
```

```python
ID_COLUMNS = ["unit_id", "cycle"]
TARGET_COLUMN = "RUL"
```

```python
def scale_dataset(train_path, val_path, train_output_path, val_output_path, scaler_output_path):
    train_df = pd.read_csv(train_path)
    val_df = pd.read_csv(val_path)

    print("Training:", train_df.shape)
    print("Validation:", val_df.shape)

    train_df = train_df.sort_values(
        by=["unit_id", "cycle"]
    ).reset_index(drop=True)

    val_df = val_df.sort_values(
        by=["unit_id", "cycle"]
    ).reset_index(drop=True)

    feature_columns = [
        column
        for column in train_df.columns
        if column not in ID_COLUMNS + [TARGET_COLUMN]
    ]

    print("\nFeature columns:\n", feature_columns)
    print("\nNumber of features:", len(feature_columns))

    scaler = MinMaxScaler()
    train_df[feature_columns] = scaler.fit_transform(train_df[feature_columns])
    val_df[feature_columns] = scaler.transform(val_df[feature_columns])

    train_df.to_csv(train_output_path, index=False)
    val_df.to_csv(val_output_path, index=False)

    joblib.dump(scaler, scaler_output_path)

    print("\nScaled dataset shapes:")
    print("Training:", train_df.shape)
    print("Validation:", val_df.shape)

    print("\nExample scaled training data:")
    print(train_df.head())

    print("\nExample scaled validation data:")
    print(val_df.head())

    print("\nScaler saved to:")
    print(scaler_output_path)

    print("\nTraining dataset saved to:")
    print(train_output_path)

    print("\nValidation dataset saved to:")
    print(val_output_path)
```

```python
scale_dataset(
    train_path=WITHOUT_CONSTANT_TRAIN,
    val_path=WITHOUT_CONSTANT_VAL,

    train_output_path=WITHOUT_CONSTANT_TRAIN_SCALED,
    val_output_path=WITHOUT_CONSTANT_VAL_SCALED,

    scaler_output_path=WITHOUT_CONSTANT_SCALER
)
```

```python
scale_dataset(
    train_path=WITH_CONSTANT_TRAIN,
    val_path=WITH_CONSTANT_VAL,

    train_output_path=WITH_CONSTANT_TRAIN_SCALED,
    val_output_path=WITH_CONSTANT_VAL_SCALED,

    scaler_output_path=WITH_CONSTANT_SCALER
)
```
