```python
import pandas as pd
```

```python
INPUT_PATH_1 = "../CMAPSSData/processed/train_FD001_cleaned_added_RUL(not_removed_the_sensor).csv"
INPUT_PATH_2 = "../CMAPSSData/processed/train_FD001_cleaned_added_RUL(removed_the_sensor).csv"
```

```python
df1 = pd.read_csv(INPUT_PATH_1)
df2 = pd.read_csv(INPUT_PATH_2)
```

```python
df1 = df1.sort_values(
    by=["unit_id", "cycle"]
).reset_index(drop=True)

df2 = df2.sort_values(
    by=["unit_id", "cycle"]
).reset_index(drop=True)
```

```python
engine_ids_df1 = sorted(df1["unit_id"].unique())

print("Total number of engines:", len(engine_ids_df1))
```

```python
engine_ids_df2 = sorted(df2["unit_id"].unique())

print("Total number of engines:", len(engine_ids_df2))
```

```python
train_ids_df1 = engine_ids_df1[:80]
val_ids_df1 = engine_ids_df1[80:]
print("\nTraining engine IDs:")
print(train_ids_df1)

print("\nValidation engine IDs:")
print(val_ids_df1)
```

```python
train_ids_df2 = engine_ids_df2[:80]
val_ids_df2 = engine_ids_df2[80:]
print("\nTraining engine IDs:")
print(train_ids_df2)

print("\nValidation engine IDs:")
print(val_ids_df2)
```

```python
train_df1 = df1[
    df1["unit_id"].isin(train_ids_df1)
].copy()

train_df2 = df2[
    df2["unit_id"].isin(train_ids_df2)
].copy()
```

```python
val_df1 = df1[
    df1["unit_id"].isin(val_ids_df1)
].copy()

val_df2 = df2[
    df2["unit_id"].isin(val_ids_df2)
].copy()
```

```python
train_df1 = train_df1.sort_values(
    by=["unit_id", "cycle"]
).reset_index(drop=True)

val_df1 = val_df1.sort_values(
    by=["unit_id", "cycle"]
).reset_index(drop=True)

train_df2 = train_df2.sort_values(
    by=["unit_id", "cycle"]
).reset_index(drop=True)

val_df2 = val_df2.sort_values(
    by=["unit_id", "cycle"]
).reset_index(drop=True)
```

```python
print("Training engines:", train_df1["unit_id"].nunique())
print("Validation engines:", val_df1["unit_id"].nunique())

print("Training rows:", len(train_df1))
print("Validation rows:", len(val_df1))

print("\n")
print("Training engines:", train_df2["unit_id"].nunique())
print("Validation engines:", val_df2["unit_id"].nunique())

print("Training rows:", len(train_df2))
print("Validation rows:", len(val_df2))
```

```python
common_engines_df1 = set(train_df1["unit_id"]) & set(
    val_df1["unit_id"]
)

print("Common engines between train and validation:")
print(common_engines_df1)

common_engines_df2 = set(train_df2["unit_id"]) & set(
    val_df2["unit_id"]
)

print("Common engines between train and validation:")
print(common_engines_df1)
```

```python
assert len(train_ids_df1) == 80
assert len(val_ids_df1) == 20

assert len(common_engines_df1) == 0

print("No engine overlap detected.")
```

```python
assert len(train_ids_df2) == 80
assert len(val_ids_df2) == 20

assert len(common_engines_df2) == 0

print("No engine overlap detected.")
```

```python
train_path_df1 = "../CMAPSSData/processed/train_FD001_cleaned_added_RUL_training(not_removed_the_sensor).csv"
val_path_df1 = "../CMAPSSData/processed/train_FD001_cleaned_added_RUL_validation(not_removed_the_sensor).csv"

train_df1.to_csv(
    train_path_df1,
    index=False
)

val_df1.to_csv(
    val_path_df1,
    index=False
)
```

```python
train_path_df2 = "../CMAPSSData/processed/train_FD001_cleaned_added_RUL_training(removed_the_sensor).csv"
val_path_df2 = "../CMAPSSData/processed/train_FD001_cleaned_added_RUL_validation(removed_the_sensor).csv"

train_df2.to_csv(
    train_path_df2,
    index=False
)

val_df2.to_csv(
    val_path_df2,
    index=False
)
```
